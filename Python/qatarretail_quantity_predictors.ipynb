```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("C:/Users/HP/Downloads/qatarretail.xlsx")
```

```python
df.head(10)
```

```python
df.info()
```

```python
#find out the nulls in each column and their reasons
nullvalues = pd.DataFrame ({
    'null values' : df.isnull().sum(),
    'null %' : (df.isna().sum() / len(df) * 100).round(2)
})

nullvalues
```

```python
#null values in payment method section are treated as cash payments
df['Payment Method'] = df['Payment Method'].fillna('Cash')

#city,product,salesperson null values are filled as unknown
df['City'] = df['City'].fillna('Unknown')
df['Salesperson'] = df['Salesperson'].fillna('Unknown')
df['Product'] = df['Product'].fillna('Unknown')
```

```python
df.info()
```

```python
#discount null values are filled with 0, assuming no discount is given on those dates or products
df.loc[df['Discount'].isnull()][['Order Date', 'Product']]
```

```python
df['Discount'] = df['Discount'].fillna('0')
```

```python
df.info()
```

```python
df = df.sort_values('Order Date').reset_index(drop = True)
```

```python
#learning demand patterns over given conditions
x = df[['City', 'Product', 'Category', 'Unit Price', 'Discount', 'Sales Channel', 'Customer Type']]
y = df['Quantity']
```

```python
from xgboost import XGBRegressor
```

```python
xgb_model = XGBRegressor(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    random_state=42
)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_cols = [
    'City',
    'Product',
    'Category',
    'Sales Channel',
    'Customer Type'
]

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ],
    remainder='passthrough'
)

x_encoded = preprocessor.fit_transform(x)

x_encoded.shape
```

```python
test_size = int(x_encoded.shape[0] * 0.2)

x1_train = x_encoded[:-test_size]
x1_test = x_encoded[-test_size:]

y1_train = y.iloc[:-test_size]
y1_test = y.iloc[-test_size:]
```

```python
xgb_model.fit(x1_train, y1_train)
```

```python
y_pred = xgb_model.predict(x1_test)
```

```python
predicted = pd.DataFrame({
    'actual values' : y1_test.values,
    'xgb pred' : y_pred
    })
predicted
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mae = mean_absolute_error(y1_test, y_pred)
mse = mean_squared_error(y1_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y1_test, y_pred)
```
