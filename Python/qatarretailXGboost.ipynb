```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("C:/Users/HP/Downloads/qatarretail.xlsx")
```

```python
df.head(10)
```

```python
df[['Order Date', 'Revenue', 'Profit', 'Quantity', 'Discount']]
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
```

```python
df['Order Date']
```

```python
monthly = (
    df.groupby(df['Order Date'].dt.to_period('M')).agg(
    Revenue = ('Revenue', 'sum'),
    Quantity = ('Quantity', 'sum'),
    Profit = ('Profit', 'sum'),
    Discount = ('Discount', 'mean')
)
.reset_index() 
)
```

```python
monthly
```

```python
monthly['Last Month'] = monthly['Revenue'].shift(1)
monthly['2 months ago'] = monthly['Revenue'].shift(2)
monthly['3 months ago'] = monthly['Revenue'].shift(3)

```

```python
monthly.head()
```

```python
monthly['roll_avg'] = (
    monthly['Revenue']
    .shift(1)
    .rolling(3)
    .mean()
)
```

```python
monthly.head()
```

```python
monthly = monthly.dropna()
```

```python
monthly.head()
```

```python
x = monthly[['Last Month', '2 months ago', '3 months ago', 'roll_avg']]
y = monthly['Revenue']
```

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X_train = x.iloc[:-4]
X_test = x.iloc[-4:]

y_train = y.iloc[:-4]
y_test = y.iloc[-4:]
```

```python
model = LinearRegression()

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print(y_pred)
```

```python
comparison = pd.DataFrame({
    'Actual Revenue': y_test.values,
    'Predicted Revenue': y_pred
})

comparison
```

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R²:", r2)
```
